```python
# To take the equation from user and convert it,
# into the agumented matrix and find rank and solution

from sympy import*
from IPython.display import display,Math

# take equation and convert it into matrix and print
m=int(input("Enter the no. of equation in given system:"))
var=symbols(input("Enter the variables(use comma seperated):"));
Equations=[] # Make a storage for equation
for i in range(1,m+1):
  raw_eq=input(f"Enter the euation{i} (like 2*x+3*y+z=d)")
  lhs,rhs=raw_eq.split("=")
  eq=Eq(sympify(lhs),sympify(rhs))
  Equations.append(eq)

display(Equations)
for eq in Equations:
  display(eq)
A,B=linear_eq_to_matrix(Equations,var)
display(Math(r"A="+latex(A)))
display(Math(r"B="+latex(B)))

# To find the agumented matrix
A_B=A.row_join(B)
display(Math(r"[A|B]="+latex(A_B)))

# To convert the matrix into row echelon form
REF=A_B.echelon_form()
display(Math(r"REF="+latex(A_B)))

# Rank of matrix B and Agumented matrix
r1=A_B.rank()
r2=A.rank()
print("r(A|B)=",r1)
print("r(A)=",r2)

# Finding the solution of equations
n=A.cols
if r1==r2==n:
  print("System is consistant and has unique solution.")
  solution=A.LUsolve(B)
  print("solution:",solution)
  for i in range(n):
      print(f"{var[i]}={solution[i]}")

elif r1==r2<n:
  print("System is consistent and it has infinite solution.")
  vars=symbols(f"x1:{n+1}")
  sol=linsolve((A,B),*vars)
  print("infinite solution:",sol)

else:
  print("System is not consistent and it does not have any solution.")




```
