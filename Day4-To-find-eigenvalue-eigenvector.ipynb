```python
from sympy import*
from IPython.display import display,Math

# take a matrix from user
m=int(input("Enter the no. of rows or columns in matrix:"))
A=zeros(m,m)
for i in range(m):
  for j in range(m):
    A[i,j]=int(input(f"Enter the matrix A{i+1}{j+1}:"))
display(Math(r"A="+latex(A)))

# To Find the eigen value and vectors
data=A.eigenvects()
display(data)

for eigenvalue,multiplicity,vectors in data:# multiplicity is a eigen value for repeted eigen vector
    print(f"Eigen value is {eigenvalue} (Repeated{multiplicity}times)")
    for k,vector in enumerate(vectors):
      print(f"Eigen vector correspond to eigen value is:")
      display(Math(latex(vector)))
```
