```python
# To find the vectors are independant or not and
# finding the relation between them
from sympy import*
from IPython.display import display,Math

# To take vectors from user
var=symbols(input("Enter the name of vectors:").split(","))
n=len(var)
vect=[]

for i in range(1,n+1):
  raw_vect=input(f"Enter the vector{i}(like[1,2,3]):")
  vect.append(eval(raw_vect))
A=Matrix(vect).T # To convert it into a transpose of matrix so the vectors convert into matrix form
display(Math(r"A="+latex(A)))

r=A.rank()
print("Rank of the matrix A:",r,"and no. of vectors are:",n)

if r==n:
  print("vectors are linearly independant")
else:
  print("vectors are linearly dependant")
# To find the relation
ns=A.nullspace()
print("Relation is:")
for vec in ns:
  eq=0
  for i in range(n):
    eq=eq+vec[i]*var[i]
  display(Math(latex(Eq(eq,0))))
  print(f"{eq}=0")

```
