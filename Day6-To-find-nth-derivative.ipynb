```python
# nth Order Derivative
from sympy import*
from IPython.display import display,Math
x=symbols('x')
y=simplify(input("Enter the function:"))
n=int(input("Enter the value of n:"))
a=int(input("Enter the constant replace to x:"))
for i in range(n):
  d=diff(y,x,i)
  display(Math(f"y{i}="+latex(d)))
  s=d.subs(x,a)
  display(Math(f"y{i}(x={a})="+latex(s)))
```
