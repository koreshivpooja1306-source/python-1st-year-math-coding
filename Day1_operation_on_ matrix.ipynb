```python
# To taka the matrix from user and perfrom operation on it.
from sympy import*
from IPython.display import display,Math

# Take matrixes from user
m=int(input("Enter the number of rows or columns for matrix A and B:"))
A=zeros(m,m)
print("Enter the enteries of matrix A ")
for i in range(m):
  for j in range(m):
    A[i,j]=input(f"Enter the matrix A{i+1}{j+1}:")
n=m
B=zeros(n,n)
print("Enter the enteries of matrix B ")
for i in range(n):
  for j in range(n):
    B[i,j]=(input(f"Enter the matrix B{i+1}{j+1}"))

display(Math(r"A="+latex(A)))
display(Math(r"B="+latex(B)))
# Addition of matrix subtraction and multiplication

display(Math(r"A+B="+latex(A+B)))
display(Math(r"A-B="+latex(A-B)))
display(Math(r"A*B="+latex(A*B)))

# Determinant and inverse of matrix by checking it is singular or not.
display(Math(r"determinant of A="+latex(A.det())))
display(Math(r"determinant of matrix b="+latex(B.det())))

if(A.det()==0):
  print("inverse of A can not find because A is singular matrix.")
if(B.det()==0):
  print("inverse of B can not find because B is singular matrix.")
if(A.det()!=0):
  print("inverse of A is possible because A is not singular matrix.")
  display(Math(r"A^{-1}="+latex(A.inv())))
if(B.det()!=0):
  print("inverse of B is possible because A is not singular matrix.")
  display(Math(r"B^{-1}="+latex(B.inv())))





```
