```python
# To find the limit of the function
# infinity symbol is double o here i.e.oo
from sympy import*
from IPython.display import display,Math
x=symbols('x')
f=simplify(input("Enter the function:"))
a=(input("Enter the limit tends to what:"))
L=limit(f,x,a)
print(f"The value of limit of{f} as a x tends to {a} is {L}")

```
