```python
# TO find the series expansion of given function
from sympy import*
from IPython.display import display,Math
x=symbols('x')
f=simplify(input("Enter the function:"))
a=int(input("Enter the point about which you are going to find series:"))
n=int(input("Enter the maximum power upto which series wants:"))
s=series(f,x,a,n+1)
pprint(s)
```
